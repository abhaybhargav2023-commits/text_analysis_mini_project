# Text Analysis

This project analyzes a sample paragraph using Python.

The program calculates:
- Number of words
- Number of characters
- Number of vowels
- Repeated words

Basic Python text processing techniques such as `split()`, `len()`, loops, dictionaries, and `Counter` are used.

```python
# Sample paragraph

text = """
Python is a popular programming language. Python is easy to learn and Python
is widely used for data analysis, machine learning, and automation.
"""

print(text)
```

```python
# Split the text into words

words = text.split()

print("Words:", words)
print("Number of Words:", len(words))
```

```python
# Count characters

character_count = len(text)

print("Number of Characters:", character_count)
```

```python
# Count vowels

vowels = "aeiou"
vowel_count = 0

for character in text.lower():
    if character in vowels:
        vowel_count += 1

print("Number of Vowels:", vowel_count)
```

```python
from collections import Counter

# Convert words to lowercase
clean_words = [word.strip(".,!?;:\"'()").lower() for word in words]

# Count word frequency
word_counts = Counter(clean_words)

# Find repeated words
repeated_words = {
    word: count
    for word, count in word_counts.items()
    if count > 1
}

print("Repeated Words:", repeated_words)
```

```python
# Final Text Analysis

print("===== TEXT ANALYSIS =====")
print("Number of Words:", len(words))
print("Number of Characters:", character_count)
print("Number of Vowels:", vowel_count)
print("Repeated Words:", repeated_words)
```
